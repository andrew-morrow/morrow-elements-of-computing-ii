# **1. Revenue & Financial KPIs**


**Total Revenue per Facility:**

Summing booking slots multiplied by facility costs (guest vs. member rates) to determine which facilities generate the most income.

```python
df_rev_per = pd.read_sql("""SELECT f.name AS facility_name, SUM(
                         CASE 
                            WHEN b.memid = 0 THEN f.guestcost * b.slots
                            ELSE f.membercost * b.slots
                         END) AS total_revenue
                         FROM facilities as f
                         JOIN bookings as b ON b.facid = f.facid
                         GROUP BY f.name
                         ORDER BY total_revenue DESC;
""", conn)

print("Total Revenue per Facility")
display(df_rev_per)
```

**Average Revenue per Member (ARPM):**

Calculate the total spending (fees + booking costs) per member to identify high-value users.

```python
df_avg_rev = pd.read_sql("""SELECT m.firstname ||' '|| m.surname AS member_name, SUM (b.slots * f.membercost) AS total_spending
                         FROM members as m
                         JOIN bookings as b ON b.memid = m.memid
                         JOIN facilities as f ON f.facid = b.facid
                         WHERE m.memid > 0
                         GROUP BY member_name
                         ORDER BY total_spending DESC
                         ;
""", conn)

print("Total Spending per Member")
display(df_avg_rev)
```

**Guest Fee Revenue:**
Total income generated specifically from non-member guests, indicating the popularity of the club with outside visitors.

```python
df_guest_rev = pd.read_sql("""SELECT m.firstname ||' '|| m.surname AS guest_name, SUM (b.slots * f.guestcost) AS total_spending
                         FROM members as m
                         JOIN bookings as b ON b.memid = m.memid
                         JOIN facilities as f ON f.facid = b.facid
                         WHERE m.memid = 0
                         GROUP BY guest_name
                         ;
""", conn)

print("Total Spending for Guests")
display(df_guest_rev)
```

**Facility Cost Recovery:**
Comparing the revenue generated by a facility to the frequency of its use to determine if rates need adjustment.

```python
df_fac_cost = pd.read_sql("""SELECT f.name AS facility_name, SUM(b.slots) AS total_uses, SUM(
                         CASE 
                            WHEN b.memid = 0 THEN f.guestcost * b.slots
                            ELSE f.membercost * b.slots
                         END) AS total_revenue
                         FROM bookings as b
                         JOIN facilities as f ON f.facid = b.facid
                         GROUP BY f.name
                         ORDER BY total_uses DESC, total_revenue DESC
                         ;
""", conn)

print("Total Spending for Guests")
display(df_fac_cost)
```
